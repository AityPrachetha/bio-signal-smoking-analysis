```python
for col in df.columns:
    if df[col].dtype != 'object':
        sns.boxplot(df[col])
        plt.xlabel('col')
        plt.show()
```

```python
list=['age','height(cm)','weight(kg)','waist(cm)','eyesight(left)','eyesight(right)','systolic','relaxation','fasting blood sugar','Cholesterol','triglyceride','HDL','LDL','hemoglobin','serum creatinine','AST','ALT','Gtp']
```

```python
for col in list:
    q1=df[col].quantile(0.25)
    q2=df[col].quantile(0.75)
    IQR=q2-q1
    lb=q1-1.5*IQR
    ub=q2+1.5*IQR
    df=df[(df[col]>=lb) & (df[col]<=ub)]
```

```python
df=pd.get_dummies(df,drop_first=True)
df = df.astype(int)
```

```python
df.to_csv("smoking_updated.csv", index=False)
```

```python
x = df.drop("smoking", axis=1)
y = df["smoking"]
```

```python
from sklearn.ensemble import ExtraTreesClassifier

model = ExtraTreesClassifier()

model.fit(x, y);
```

```python
feature_importance = pd.Series(
    model.feature_importances_,
    index=x.columns
)
print(feature_importance.sort_values(ascending=False))
```

```python
top_features = feature_importance.nlargest(15)
print(top_features)
```

```python
x = x[top_features.index]
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

x_train = scaler.fit_transform(x_train)

x_test = scaler.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression()

lr.fit(x_train, y_train);
```

```python
y_pred = lr.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy :", accuracy)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

```python
from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier()

dt.fit(x_train, y_train);
```

```python
y_pred_dt = dt.predict(x_test)

print("Decision Tree Accuracy :",
      accuracy_score(y_test, y_pred_dt))
```

```python
from sklearn.ensemble import BaggingClassifier

bag = BaggingClassifier()

bag.fit(x_train, y_train);
```

```python
y_pred_bag = bag.predict(x_test)

print("Bagging Accuracy :",
      accuracy_score(y_test, y_pred_bag))
```

```python
from sklearn.ensemble import ExtraTreesClassifier

et = ExtraTreesClassifier()

et.fit(x_train, y_train);
```

```python
y_pred_et = et.predict(x_test)

print("Extra Trees Accuracy :",
      accuracy_score(y_test, y_pred_et))
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier()

rf.fit(x_train, y_train);
```

```python
y_pred_rf = rf.predict(x_test)

print("Random Forest Accuracy :",
      accuracy_score(y_test, y_pred_rf))
```

```python

# Logistic Regression
y_pred_lr = lr.predict(x_test)
lr_acc = accuracy_score(y_test, y_pred_lr)

# Decision Tree
y_pred_dt = dt.predict(x_test)
dt_acc = accuracy_score(y_test, y_pred_dt)

# Bagging Classifier
y_pred_bag = bag.predict(x_test)
bag_acc = accuracy_score(y_test, y_pred_bag)

# Extra Trees
y_pred_et = et.predict(x_test)
et_acc = accuracy_score(y_test, y_pred_et)

# Random Forest
y_pred_rf = rf.predict(x_test)
rf_acc = accuracy_score(y_test, y_pred_rf)

# Print All Accuracies
print("Logistic Regression Accuracy :", lr_acc * 100)

print("Decision Tree Accuracy :", dt_acc * 100)

print("Bagging Classifier Accuracy :", bag_acc * 100)

print("Extra Trees Accuracy :", et_acc * 100)

print("Random Forest Accuracy :", rf_acc * 100)
```
